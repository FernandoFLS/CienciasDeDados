## Este dataset possui 22 colunas com dados demográficos, socioeconômicos e de saúde pública. A seguir está lista das variáveis e seus respectivos significados


#Informações de Identificação e Categoria
#####Country: nome do país analisado.
#####Year: ano da coleta dos dados (varia de 2000 a 2015).
#####Status: status de desenvolvimento do país, classificado como Developing (Em desenvolvimento) ou Developed (Desenvolvido).

#Variável Alvo (Target)
#####Life expectancy: expectativa de vida média da população (em anos).

#Indicadores de Mortalidade
#####Adult Mortality: Taxa de mortalidade de adultos de ambos os sexos (probabilidade de morrer entre os 15 e 60 anos por 1.000 habitantes).
#####infant deaths: Número de mortes de bebês (menores de 1 ano) por 1.000 habitantes.
#####under-five deaths: Número de mortes de crianças menores de 5 anos por 1.000 habitantes.

#Fatores de Saúde e Estilo de Vida
#####Alcohol: Consumo per capita de álcool puro registrado em litros (para maiores de 15 anos).
#####BMI (Body Mass Index): Índice de Massa Corporal médio de toda a população do país.
#####Thinness 1-19 years: Prevalência de magreza em crianças e adolescentes de 1 a 19 anos (%) - ou baixo peso - refere-se à proporção de indivíduos em uma população que apresentam um Índice de Massa Corporal (IMC) abaixo de determinado valor
#####Thinness 5-9 years: Prevalência de magreza em crianças de 5 a 9 anos (%) - ou baixo peso - refere-se à proporção de indivíduos em uma população que apresentam um Índice de Massa Corporal (IMC) abaixo de determinado valor

#Cobertura de Imunização (Vacinas)
#####Hepatitis B: Cobertura de vacinação contra Hepatite B em crianças de 1 ano (%)
#####Measles: Número de casos relatados de Sarampo por 1.000 habitantes.
#####Polio: Cobertura de vacinação contra Pólio (Paralisia Infantil) em crianças de 1 ano (%).
#####Diphtheria: Cobertura de vacinação contra Diphtheria, Coqueluche e Tétano (DTP3) em crianças de 1 ano (%).

#Fatores Econômicos e Sociais
#####percentage expenditure: Gasto com saúde como porcentagem do Produto Interno Bruto (PIB) per capita (%).
#####Total expenditure: Gasto do governo geral com saúde como porcentagem do gasto total do governo (%).
#####GDP: Produto Interno Bruto (PIB) per capita (em dólares americanos).
#####Population: População total do país.
#####Income composition of resources: Índice de Desenvolvimento Humano (IDH) em termos de composição de renda dos recursos (varia de 0 a 1).
#####Schooling: Número médio de anos de escolaridade que uma pessoa frequenta.

#Doenças Crônicas e Infecciosas
#####HIV/AIDS: Mortes por 1.000 nascidos vivos decorrentes de HIV/AIDS (em crianças de 0 a 4 anos).

# Crosstab (tabela com duas variáveis)

```python
# Crosstab é como um pivot table fazendo uma contagem dos registros entre linhas e colunas.


# 1. Primeiro, definimos a lista com os países que você quer analisar
paises_selecionados = ['Brazil', 'Argentina', 'Chile', 'Uruguay']

# 2. Depois, criamos o df_paises filtrando o DataFrame original (df_menor)
df_paises = df_menor.loc[df_menor['Country'].isin(paises_selecionados)]

# 3. Cria a tabela cruzada de País por Ano calculando a média da expectativa de vida
tabela_cruzada = pd.crosstab(
    index=df_paises['Year'],
    columns=df_paises['Country'],
    values=df_paises['Life expectancy'],
    aggfunc='mean'
)

tabela_cruzada
```

```python
# 1. Cria a paleta de cores suave (ex: tons de azul claro/pastel)
# O 'as_cmap=True' transforma a paleta do Seaborn em um mapa de cores para o Pandas
cm = sns.light_palette("seagreen", as_cmap=True) # Você também pode usar "skyblue", "purple", etc.


# O que cada comanda faz:

# .light_palette("seagreen", as_cmap=True): Cria um gradiente suave que começa quase branco e vai escurecendo até um verde oceano suave conforme a expectativa de vida aumenta.
# .background_gradient(cmap=cm): Pinta o fundo das células com base nos valores (valores maiores ganham cores ligeiramente mais fortes).
# .format(precision=1): Arredonda todos os números para apenas uma casa decimal (ex: 75.4 em vez de 75.3892)
# .set_properties e set_table_styles: Alinha os números no centro da célula e deixa os cabeçalhos com um tom cinza bem moderno e limpo.


# 2. Aplica a formatação na tabela cruzada
tabela_final = tabela_cruzada.style\
    .background_gradient(cmap=cm)\
    .format(precision=1)\
    .set_properties(**{'text-align': 'center', 'font-size': '12px'})\
    .set_table_styles([{'selector': 'th', 'props': [('background-color', '#f4f4f4'), ('color', '#333')]}])

# Exibe a tabela formatada
tabela_final
```
